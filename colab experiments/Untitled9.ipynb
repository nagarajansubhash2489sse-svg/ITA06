```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

# Create sample data
np.random.seed(42)

# Generate two groups of data
data1 = np.random.normal(loc=2, scale=0.7, size=100)
data2 = np.random.normal(loc=7, scale=0.8, size=100)

# Combine the data
X = np.concatenate([data1, data2]).reshape(-1, 1)

# Create Gaussian Mixture Model
# EM algorithm is used internally by GaussianMixture
model = GaussianMixture(
    n_components=2,
    random_state=42
)

# Train the model using EM algorithm
model.fit(X)

# Predict cluster for each data point
labels = model.predict(X)

# Display results
print("EM Algorithm - Gaussian Mixture Model")
print("--------------------------------------")

print("\nMeans of the Gaussian Components:")
print(model.means_.flatten())

print("\nVariances of the Gaussian Components:")
print(model.covariances_.flatten())

print("\nMixing Weights:")
print(model.weights_)

print("\nFirst 20 Cluster Assignments:")
print(labels[:20])

print("\nLog-Likelihood:")
print(model.score(X))

# Plot the clusters
plt.figure(figsize=(8, 5))

plt.scatter(
    X[:, 0],
    np.zeros(len(X)),
    c=labels,
    s=50
)

plt.xlabel("Data Value")
plt.ylabel("Cluster")
plt.title("Expectation-Maximization Algorithm")
plt.grid(True)

plt.show()
```
