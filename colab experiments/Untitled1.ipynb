```python
import pandas as pd

data = {
    'Sky':['Sunny','Sunny','Rainy','Sunny'],
        'AirTemp':['Warm','Warm','Cold','Warm'],
            'Humidity':['Normal','High','High','High'],
                'Wind':['Strong','Strong','Strong','Strong'],
                    'Water':['Warm','Warm','Warm','Cool'],
                        'Forecast':['Same','Same','Change','Change'],
                            'EnjoySport':['Yes','Yes','No','Yes']
                            }

df = pd.DataFrame(data)

# Save as CSV
df.to_csv("trainingdata.csv", index=False)

print("CSV file created successfully!")
```

```python
import pandas as pd
import numpy as np

# Read the CSV file
data = pd.read_csv("trainingdata.csv")

print("Training Dataset:\n")
print(data)

# Separate attributes and target
concepts = np.array(data.iloc[:, :-1])
target = np.array(data.iloc[:, -1])

# Initialize Specific Hypothesis
S = concepts[0].copy()

# Initialize General Hypothesis
G = [['?' for i in range(len(S))] for i in range(len(S))]

print("\nInitial Specific Hypothesis:")
print(S)

print("\nInitial General Hypothesis:")
for row in G:
    print(row)

# Candidate Elimination Algorithm
for i, h in enumerate(concepts):

    if target[i] == "Yes":

        for x in range(len(S)):
            if h[x] != S[x]:
                S[x] = '?'
                G[x][x] = '?'

    elif target[i] == "No":

        for x in range(len(S)):
            if h[x] != S[x]:
                G[x][x] = S[x]
            else:
                G[x][x] = '?'

print("\nFinal Specific Hypothesis:")
print(S)

print("\nFinal General Hypothesis:")

G_final = []

for row in G:
    if row != ['?', '?', '?', '?', '?', '?']:
        G_final.append(row)

for row in G_final:
    print(row)
```
