```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder

# Create Label Encoders
le = LabelEncoder()

# Encode all columns
df_encoded = df.copy()

for column in df.columns:
    df_encoded[column] = le.fit_transform(df[column])

print("\nEncoded Dataset")
print(df_encoded)

# Features and Target
X = df_encoded.iloc[:, :-1]
y = df_encoded.iloc[:, -1]

# Train ID3 Decision Tree
model = DecisionTreeClassifier(criterion='entropy')

model.fit(X, y)

print("\nDecision Tree Model Trained Successfully")

# New Sample
new_sample = [['Sunny','Cool','High','Strong']]

# Encode the new sample
sample = []

for i, col_name in enumerate(df.columns[:-1]):
    encoder_for_feature = LabelEncoder()
    encoder_for_feature.fit(df[col_name])
    sample.append(encoder_for_feature.transform([new_sample[0][i]])[0])

prediction = model.predict([sample])

# Decode Prediction
target_column_name = df.columns[-1]
target_encoder_for_decoding = LabelEncoder()
target_encoder_for_decoding.fit(df[target_column_name])
decoded_prediction_label = target_encoder_for_decoding.inverse_transform(prediction)[0]

print("\nNew Sample:")
print(new_sample[0])

print("\nPrediction:")
print("Play Tennis =", decoded_prediction_label)
```
